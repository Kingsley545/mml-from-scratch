```python
def dot_product(A, B):
    """returns the dot product between two matrices"""
    assert len(A) == len(B), "both matrices must have the same length"
    return sum([i * j for i, j in zip(A, B)])
```

```python
def transpose(A):
    """ returns the transpose of the matrix A """
    m, n = len(A), len(A[0])
    result = [[0 for _ in range(m)] for _ in range(n)]
    for i in range(n):
        for j in range(m):
            result[i][j] = A[j][i]
    return result
```

```python
def mat_mul(A, B):
    """returns the matrix multiplication between two matrices"""
    m1, n1 = len(A), len(A[0])
    m2, n2 = len(B), len(B[0])
    assert n1 == m2, "number of columns of first matrix must equal the number of rows of second matrix"
    result = [[0 for _ in range(n2)] for _ in range(m1)]
    for i in range(m1):
        for j in range(n2):
            result[i][j] = dot_product(A[i], transpose(B)[j])

    return result
```
