```python
import numpy as np
```

```python
def dot_product(A, B):
    """returns the dot product between two vectors"""
    assert len(A) == len(B), "both vectors must contain the same number of elements"
    return sum([i * j for i, j in zip(A, B)])
```

```python
def transpose(A):
    """ returns the transpose of the matrix A """
    m, n = len(A), len(A[0])
    result = [[0 for _ in range(m)] for _ in range(n)]
    for i in range(n):
        for j in range(m):
            result[i][j] = A[j][i]
    return result
```

```python
def mat_mul(A, B):
    """returns the matrix multiplication between two matrices"""
    m1, n1 = len(A), len(A[0])
    m2, n2 = len(B), len(B[0])
    assert n1 == m2, "number of columns of first matrix must equal the number of rows of second matrix"
    result = [[0 for _ in range(n2)] for _ in range(m1)]
    Bt = transpose(B)
    for i in range(m1):
        for j in range(n2):
            result[i][j] = dot_product(A[i], Bt[j])

    return result
```

```python
rng = np.random.default_rng(42)

for _ in range(20): 
    m, n, p = rng.integers(1, 6, 3)
    X = rng.integers(-5, 5, (m, n))
    Y = rng.integers(-5, 5, (n, p))
    assert np.allclose(mat_mul(X.tolist(), Y.tolist()), X@Y)
    assert np.allclose(transpose(X.tolist()), X.T)
    assert np.isclose(dot_product(X[0].tolist(), X[0].tolist()), X[0]@X[0])
    
print("all tests passed")
```
